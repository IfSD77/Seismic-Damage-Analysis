# Predicting Seismic Building Damage in the 2015 Gorkha Earthquake

**A Data Science Analysis of Structural Vulnerability and Regional Ground Motion Intensity**

SoftUni Data Science Course Project
November–December 2026

## 1. Introduction

On 25 April 2015, a magnitude 7.8 earthquake struck near Gorkha, Nepal. The event caused widespread destruction, resulted in nearly 9,000 fatalities, and damaged or destroyed hundreds of thousands of buildings. In the aftermath, one of the largest post-disaster building surveys ever conducted was carried out by Kathmandu Living Labs and the Central Bureau of Statistics of Nepal. The resulting dataset contains detailed structural characteristics and observed damage grades for more than 260,000 buildings.

This project uses that survey data, combined with regional seismic intensity information, to investigate the relationship between building attributes and observed damage. The primary goal is not simply to achieve high predictive accuracy, but to develop an interpretable model that can highlight which structural features most strongly influence seismic vulnerability.

From a civil engineering perspective, such insights can support prioritization of retrofitting efforts and inform discussions about construction practices in seismically active regions.

## 2. Problem Formulation

### 2.1 Problem Statement

Given pre-earthquake structural characteristics of a building and an estimate of the local ground motion intensity, can we predict the ordinal damage grade that the building sustained, and more importantly, can we identify which building features contribute most to higher damage levels?

### 2.2 Research Questions

1. Which structural and material characteristics are the strongest predictors of building damage grade?
2. How does regional seismic intensity interact with building vulnerability?
3. Can an interpretable machine learning approach provide actionable insights for assessing seismic risk at the building level?
4. What are the main limitations of predicting damage using only pre-event building attributes and regional intensity measures?

### 2.3 Scope and Assumptions

- The analysis is limited to buildings surveyed after the 2015 Gorkha earthquake.
- Damage grade is treated as an ordinal classification problem (grades 1–3 in the primary dataset).
- Ground motion intensity is represented at a regional (ward or district) level rather than at the individual building location.
- The models do not incorporate detailed structural engineering calculations (e.g., capacity curves or nonlinear dynamic analysis).
- Results are intended for exploratory and prioritization purposes, not for official risk assessment or design decisions.

## 3. Data Sources

This project uses two independent data sources.

### 3.1 Source 1: Building Characteristics and Damage Grades (2015 Gorkha Earthquake)

The primary dataset comes from the post-earthquake building survey conducted after the 25 April 2015 Gorkha earthquake in Nepal. It contains structural attributes and observed damage grades for hundreds of thousands of buildings. Key groups of variables include:

- Geographic identifiers (nested administrative levels)
- Building geometry (number of floors, age, area and height percentages)
- Structural system and materials (foundation type, roof type, superstructure materials)
- Occupancy and ownership information
- Observed damage grade (target variable)

This dataset is publicly available and has been widely used in data science competitions and research (commonly known as the Richter’s Predictor dataset).

### 3.2 Source 2: Regional Seismic Intensity

The second source provides estimates of ground motion intensity (e.g., Modified Mercalli Intensity or related intensity measures) at a regional level for the same earthquake. These data allow us to incorporate the strength of shaking as an additional explanatory factor, rather than relying solely on building attributes.

Combining both sources enables the analysis of how structural characteristics interact with varying levels of ground motion intensity.

## 4. Objectives

The main objectives of this project are:

1. **Data Integration**
   Combine building-level structural data with regional seismic intensity information into a coherent analytical dataset.

2. **Exploratory Analysis**
   Investigate the distribution of damage grades and their relationship with key structural characteristics (materials, number of floors, age, foundation type, etc.) and ground motion intensity.

3. **Predictive Modeling**
   Develop and compare classification models that predict the ordinal damage grade of buildings based on pre-earthquake attributes and regional intensity.

4. **Interpretability**
   Identify the most influential features driving higher damage levels using model-agnostic and model-specific interpretation techniques.

5. **Engineering Insights**
   Translate the data-driven findings into practical observations relevant to seismic vulnerability assessment, while clearly stating the limitations of the approach.
